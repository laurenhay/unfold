```python

outputs = [ pkl.load(open("../coffeaOutput/dijet/dijetHistsTest_addJMRJMSplots_JetHT_2016APV_newmatch2.5.pkl", "rb")),
            pkl.load(open("../coffeaOutput/dijet/dijetHistsTest_addJMRJMSplots_JetHT_2016_newmatch2.5.pkl", "rb")),
            pkl.load(open("../coffeaOutput/dijet/dijetHistsTest_addJMRJMSplots_JetHT_2017_newmatch2.5.pkl", "rb")),
            pkl.load(open("../coffeaOutput/dijet/dijetHistsTest_addJMRJMSplots_JetHT_2018_newmatch2.5.pkl", "rb"))
          ]

bins = binning.bin_edges(True)
hist_keys = ['ptjet_rhojet_u_reco', 'ptjet_rhojet_g_reco']
out_dict = {}
for i, output in enumerate(outputs):
    for key in hist_keys:
        if i == 0:
            rebinned_output = rebin_hist(output[key], 'ptreco', bins.pt_edges)
            rebinned_output = rebin_hist(rebinned_output, 'mpt_reco', bins.rho_edges)
            out_dict[key] = rebinned_output.project('ptreco', 'mpt_reco', 'systematic')
        else:
            rebinned_output = rebin_hist(output[key], 'ptreco', bins.pt_edges)
            rebinned_output = rebin_hist(rebinned_output, 'mpt_reco', bins.rho_edges)
            out_dict[key] += rebinned_output.project('ptreco', 'mpt_reco', 'systematic')

pkl.dump(out_dict, open("./inputs/rho/rho_data_all.pkl", "wb"))
```
